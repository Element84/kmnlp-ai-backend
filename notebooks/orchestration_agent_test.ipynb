```python
from e84_geoai_common.llm.core import (
    LLMMessage,
    LLMToolResultContent,
)
from e84_geoai_common.llm.models.claude import (
    CLAUDE_BEDROCK_MODEL_IDS,
    BedrockClaudeLLM,
)
from rich import print as rich_print

from llm_agent.agents.orchestration_agent.orchestration_agent import (
    OrchestrationAgent,
)
from llm_agent.tools.data_analysis_tool import (
    data_analysis_tool,
)
from llm_agent.tools.data_search_tool import (
    data_search_tool,
)
```

```python
import base64
from PIL import Image
import io
from matplotlib import pyplot as plt
import numpy as np
```

## Example #1

```python
llm = BedrockClaudeLLM(
    model_id=CLAUDE_BEDROCK_MODEL_IDS["Claude 3.5 Sonnet"]
)
agent = OrchestrationAgent(
    llm, tools=[data_search_tool, data_analysis_tool]
)
```

```python
responses = agent.run(
    LLMMessage(
        content="What was the mean sea surface temperature in the Black Sea on June 01, 2024?",
    )
)
for response in responses:
    rich_print(response)
    print("------------------------------------")
```
