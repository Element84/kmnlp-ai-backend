```python
import os

import h5netcdf
import netCDF4
```

```python
file_name = f"{os.getenv("HOME")}/Downloads/CMORPH_V1.0_ADJ_8km-30min_2024083123.nc"
```

```python
nc_file = h5netcdf.File(file_name, 'r')

with nc_file as f:
    # Get dimensions
    dimensions = f.dimensions
    variables = f.variables
    # Print dimensions
    print(dimensions)
    print(variables)

    # Iterate through dimensions
    for dim_name, dim_size in dimensions.items():
        print(f"Dimension: {dim_name}, Size: {dim_size}")

    for var_name, variable in variables.items():
        print(f"Variable: {var_name}, {variable}")
```

# How to use

NetCDF files may have different attributes in them. The cell below prints out a bunch of attributes. Some are commented out because they were not present in the dataset being investigated.

To discover what attributes are available, uncomment the `print(dir(dataset))` line. Add any attributes you want to see the value of to the list below, and comment-out (or delete) any that don't exist.

```python
dataset = netCDF4.Dataset(file_name)

# print(dir(dataset))
attributes = [
        "title",
        "name",
        "summary",
        # "comment",
        "references",
        "Conventions",
        "contributor_name",
        "creator_name",
        "data_model",
        "keywords",
        "keywords_vocabulary",
        "processing_level",
        "sensor",
        "standard_name_vocabulary",
        "dimensions",
        "geospatial_bounds",
        "geospatial_lat_min",
        "geospatial_lat_max",
        "geospatial_lat_units",
        "geospatial_lat_resolution",
        "geospatial_lon_max",
        "geospatial_lon_units",
        "geospatial_lon_resolution",
        "time_coverage_duration",
        "time_coverage_end",
        "time_coverage_start",
        "spatial_resolution",
        "variables",
]

for a in attributes:
    try:
        print(f"{a}: {getattr(dataset, a)}")
    except AttributeError:
        print(f"{a}: Attribute not found.")

    print("\n")
```
